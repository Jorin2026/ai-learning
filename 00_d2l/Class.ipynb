```python
class Robot:
    def __init__(self,name:str ,battery:int =0  ,color:str | None=None,material=None,attribute : list | None =None):
        
        self.name = name 
        self.battery = battery 
        self.color =color
        self.material = material

    def manufacturer(self,producter:str=None,data:str="00/00/00"):
        self.producter = producter
        self.data = data
        print(f"{ self.producter} - {self.data}")
```
