```python
数字们 = [1,2,3,4]

#笨办法 
平方们 =[]
for n in 数字们:
    平方们.append(n *n) 

```

```python
print(平方们)
```

```python
#列表推到式（等价）:
平方们2 = [n * n for n in 数字们 ] 
print(平方们2)
```
