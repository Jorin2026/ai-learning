```python
battery = 50 
if battery <30 :
    print("电量不足， 需要充电")  #这里的if里的代码被完全跳过了， 什么都不答应， 这正是， if的核心租用， 只有条件成立才执行， 不成立就跳过
```

```python
#现在扩展else ,如果不成立就跳过


battery = 50 

if battery < 30 :
    print("电池电量不足， 需要充电")
elif battery < 80 :
    print("电量中等")
else:
    print("电池电量充足")
```

```python
levels = [10,45,90,25,100]
for battery in levels:
    if battery <30 :
        print("电量不足")
    elif battery <80 :
        print("电量中等")
    else:
        print("电量充足")
        
```
