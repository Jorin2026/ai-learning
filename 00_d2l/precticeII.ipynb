```python
class 乘法器:
    def __init__(self,倍数):
        self.倍数 = 倍数

    def __call__(self,数字):
        return 数字 * self.倍数
```

```python
翻三倍 = 乘法器(3)
翻五倍 = 乘法器(5)
print(翻三倍(10))
print(翻五倍(10))
```

```python
class 没有call的乘法器:
    def __init__(self,倍数):
        self.倍数 = 倍数
```

```python
class 盒子:
    def __init__(self,大小):
        self.大小 =大小
        计算结果 = self._算个东西(大小)
        self.翻倍值 = 计算结果 

    def _算个东西(self,大小):
        return 大小 * 2 

我的盒子 = 盒子(5)
```

```python
print(我的盒子.大小)
print(我的盒子.翻倍值)
```
