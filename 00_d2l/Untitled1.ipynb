```python
#第一步字典要解决什么问题
#self.name= "瓦力"
#self.battery = 80   #这种一个名字对应一个值的结构就是字典的核心思想， 只不过，class 是提前设计好格子的名字叫什么， 而字典可以更灵活的随时增删， 这些名字-值 的配对不需要专门的类

#第二步 创建一个字典， 
robot_info = {"name":"瓦力","battery":80,"color":"灰色"}
#语法用大括号， {} 里面每一项写"名字" : 值用逗号隔开 比如名字的专门叫法叫key 键， 对应的值比如"瓦力" 叫value 字典就是一堆键值对的组合
#跟list最大的区别就是， list是按位置（0，1，2）取值， 字典是按名字取值不需要记位置
#第三步
print(robot_info["name"])
print(robot_info["battery"])
```
