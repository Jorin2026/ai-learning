```python
%matplotlib inline
import random
import torch
from d2l import torch as d2l 
```

```python
def synthetic_data(w,b,num_examples):
    "生成Xw + b + 噪声 "
    X = torch.normal(0,1,(num_examples,len(w)))
    y = torch.matmul(X,w)+b
    y += torch.normal(0,0.001,y.shape)
    return X,y.reshape(-1,1)
```

```python
true_w = torch.tensor([2,-3.4])
true_b = 4.2
features,labels = synthetic_data(true_w,true_b,1000)
```

```python
print("features:",features[:5],"features.shape:",features.shape,"\n","labels:",labels[:5],"labels:",labels.shape)
```

```python
#d2l.set_figsize()
```

```python
def data_iter(batch_size, features,labels):
    """生成一个数据迭代器"""
    example = len(features)
    indices = list(range(example))
    print(indices[:5])
    random.shuffle(indices)
    for i in range(0,example,batch_size):
        batch_indices = torch.tensor(indices[i:min(i+batch_size,example)])
        yield features[batch_indices],labels[batch_indices]
    print("data_iter结束了")
```

```python
batch_size = 10 
data_iter(batch_size,features,labels)
```

```python
w = torch.normal(0,0.01,size=(2,1),requires_grad=True)
b = torch.zeros(1,requires_grad=True)
```

```python
print("w:",w,"b",b)
```

```python
def linreg(X,w,b):
    """线性回归模型"""
    return torch.matmul(X,w)+b
```

```python
def loss(y_hat,y): 
    """squared loss"""
    return (y_hat - y)**2/2
```

```python
def sgd(params,lr,batch_size):
    """stochastic gradient descent"""
    with torch.no_grad():
        for param in params:
            param -= lr * param.grad/batch_size
            param.grad.zero_()
            
```

```python
lr = 0.003 
num_epochs = 3
net = linreg

for epoch in range(num_epochs):
    for X,y, in data_iter(batch_size,features, labels):
        l=loss(net(X,w,b),y)
        l.sum().backward()
        sgd([w,b],lr,batch_size)
    with torch.no_grad():
        train_l = loss(net(features, w,b),labels)
        print(f'epoch{epoch+1},loss{float(train_l.mean()):f}')
       
```
