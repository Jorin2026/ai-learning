```python
skills = ["飞行","激光眼","变形"]
for skill in skills:   #for是关键字， 开始循环， skill是你自己起的临时变量名 每一轮它会一次变成skills里的每一个元素 in关键字从里面， skills你要遍历的那个list ,: 表示下面是循环体要重复执行的内容， print 
    print(skill)   #对于skills 这个list 里面的每一个东西轮流把它叫做skill, 然后执行下面的缩进的代码， for -each ， python里的for默认就是这种风格
```
