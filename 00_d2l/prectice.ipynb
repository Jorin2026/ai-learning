```python
#工具库存管理系统
#工具房里有很多种工具，每种工具都有名字和库存数量， 者两个基本信息。我们要做一个系统， 
#1.能记录每种工具现在能剩多少，
#2.支持“借出 数量减少， 归还数量增加
#3.能查询哪些工具库存不足， 这里会用到今天练的塞选框架， 能统计那种工具被借的最多， 这里会直接用上most_common
```

```python
class Tool:
    def __init__(self,name=str,quantity:int=0):
        self.name = name
        self.quantity = quantity

    def __str__(self):
        return f"{self.name},库存{self.quantity}"


    def borrow(self,borrownum):
        if borrownum > self.quantity:
            print("借数量已经超过库存内现有的数量，库内现存数量是：",self.quantity)
        else:
            self.quantity-=borrownum

    def give_back(self,backnum):
        self.quantity+=backnum
        
```

```python
斜口钳 = Tool("斜口钳",30)
print(斜口钳)
```

```python
扳手= Tool("扳手",10)
print(扳手.name)
print(扳手.quantity)
```

```python
扳手.borrow(100)
print(扳手.quantity)
```

```python
扳手.give_back(5)
print(扳手.quantity)
```

```python
def low_stock(仓库,threshold):
    清单=[]
    for 名字 in 仓库:
        该工具 = 仓库[名字]
        if 该工具.quantity < threshold:
            清单.append(名字)
    return 清单
```
