```python

def count_item(item_list):
    统计 = {}
    for i in item_list:
        if i in 统计:
            统计[i] = 统计[i] +1
        else:
            统计[i] =1
    return 统计
```

```python
items=["扳手","螺丝刀","扳手","钳子","扳手","螺丝刀"]
x = count_item(items)
print(x)
```
