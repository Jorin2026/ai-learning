```python
#字典要解决什么问题
#self.name = "瓦力"
#self.batter =80
#这种一个名字对应一个值的结构其实就是字典里最核心的思想， 只不过class是提前设计好格子的名字叫什么
#而字典可以更灵活， 随时增删这些名字-值的配对。 不需要专门写一个类 
robot_info ={ "name":"瓦力","battery":"80","color":"灰色"}
print(robot_info["name"])
print(robot_info["battery"])
```

```python
#字典加for的循环
robot_info={"name":"瓦力","battery":"80","color":"灰色"}
for key ,value in robot_info.items():
    print(key,value)
```

```python
#再拓展一个使用操作， 修改和新增字典的内容
robot_info["battery"]=50
print(robot_info)

robot_info["material"]="钛合金"
print(robot_info)
```

```python
#下面我们进入异常处理
battery = "满电量" 
try : 
    result = battery + 10 
    print(result)
except:
    print("出错了，电量格式不对")
```

```python
##拓展抓住具体是哪种错误（更精准的写法）
battery ="满电量" 
try :
    result = battery + 10 
    print(result) 
except TypeError:
    print("类型不对，电量应该是数字，不是文字")
```

```python
def add(*args):
    print(args)

add(1,2,3,4)
```

```python
def test(*args):
    print(args[0])
    print(args[1])

test(1,2,3,4,5)
```

```python
test("小明","小花","小张","小狗")
```

```python
def test(**kwarg):  #** kwarg 会把你收入的值， 变成dict 键值对
    print(kwarg)

test(name="小明",age=28,job="工程师")
```
