```python
def count_item(item_list):
    统计={}
    for i in item_list:
        if i in 统计:
            统计[i]=统计[i]+1
        else:
            统计[i]=1
    return 统计
```

```python
class Tool:
    def __init__(self,name=str,quantity:int= 0):
        self.name = name
        self.quantity =quantity
        self.borrow_times =borrow_times =0
        
    def __str__(self):
        return f"{self.name},库存{self.quantity}"
        
    def __repr__(self):
        return f"{self.name},库存{self.quantity}"

    def borrow(self,less):
        if less > self.quantity:
            print("借的数量大于仓库内所有的数量，仓库内数量为:",self.quantity)
        else:
            self.quantity -=less
            self.borrow_times +=1;

    def give_back(self,backs):
        self.quantity +=backs
        

```

```python
仓库2={}
仓库2["扳手"] =  Tool("扳手",10)
仓库2["螺丝刀"] = Tool("",5)
仓库2["钳子"]= Tool("钳子",8)
print(仓库2)      
```

```python
仓库2["扳手"].borrow(2)
仓库2["螺丝刀"].borrow(5)
仓库2["钳子"].borrow(1)
```

```python
def frequent_items(item_list,threshold):
    统计 = count_item(item_list) #将一个列表转换成字典的形式
    清单 = [] #先建立一个清单， 然后准备放我们要放的东西
    for i in 统计:
        if 统计[i] >= threshold:
            清单.append(i)

        return 清单
```

```python
def low_stock(仓库,threshold):
    #统计 = count_item(item_list)
    低于清单 = []
    for 名字 in 仓库:
        该工具 = 仓库[名字]
        if 该工具.quantity < threshold:
            低于清单.append(名字)
    return 低于清单
    
```

```python
def most_borrowed(仓库):
    最大次数 = None 
    最大名字 = ""
    
    for 名字 in 仓库:
        该工具 = 仓库[名字]    #用名字换出实例
        
        if 最大次数 is None:
            最大次数 = 该工具.borrow_times
            最大名字 = 名字
        elif 该工具.borrow_times > 最大次数 :
            最大次数 = 该工具.borrow_times
            最大名字 = 名字
    return 最大名字
                  
```
